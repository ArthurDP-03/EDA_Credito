##Comparando os clientes inativos:

SELECT

    faixa_etaria,

    escolaridade,

    tipo_cartao,

    sexo,

    status_cliente,

    COUNT(*) AS quantidade

FROM (

    SELECT

        CASE

            WHEN idade BETWEEN 18 AND 25 THEN '18-25'

            WHEN idade BETWEEN 26 AND 35 THEN '26-35'

            WHEN idade BETWEEN 36 AND 45 THEN '36-45'

            WHEN idade BETWEEN 46 AND 60 THEN '46-60'

            WHEN idade > 60 THEN '60+'

        END AS faixa_etaria,

        escolaridade,

        tipo_cartao,

        sexo,

        CASE

            WHEN meses_inativo_12m = 0 THEN
            'Ativo'

            ELSE 'Inativo'

        END AS status_cliente

    FROM credito

) AS subconsulta

GROUP BY faixa_etaria, escolaridade, tipo_cartao,

sexo, status_cliente

HAVING COUNT(*) > 85

ORDER BY faixa_etaria, escolaridade, tipo_cartao,

sexo, status_cliente;

Com esse código conseguimos carregar um CSV apartir de uma requisição dos tipos de pessoas que mais tem em quantidade. O que isso quer dizer: Que pra cada grupo de tipo de clientes que temos esses que foram selecionados temos em maior quantidade:
